<h3 style="color:blue;">1.1 Remove/treat any special characters and non-numeric entries from financial fields.</h3>

####  This cleans the financial data and keeps only numeric values for calculations
#### pd.to_numeric() converts values into numbers (int/float) so we can perform calculations

```python

df["TransactionAmount"] = df["TransactionAmount"].replace('[₹,]', '', regex=True)
df["AccountBalance"] = df["AccountBalance"].replace('[₹,]', '', regex=True)
df.head()
```

**<h3 style="color:blue;">1.2 Convert currency amounts into numerical format.</h3>**

#### Convert 'TransactionAmount' and 'AccountBalance' columns into numeric format
#### Any invalid or non-convertible values will be replaced with NaN (missing values)

```python
df["TransactionAmount" ] = pd.to_numeric(df["TransactionAmount"], errors='coerce')
df["AccountBalance"] = pd.to_numeric(df["AccountBalance"], errors='coerce')
df.head()
```

#### Checking Missing

```python
df.isnull().sum()
```

#### **Fill missing**

```python
df["TransactionAmount"] = df["TransactionAmount"].fillna(df["TransactionAmount"].median())
df["AccountBalance"] = df["AccountBalance"].fillna(df["AccountBalance"].median())
```

```python
df.head()
```

**<h3 style="color:blue;">**1.3 Validate and format date columns.</h3>**

```python
df["TransactionDate"] = pd.to_datetime(df["TransactionDate"], format='%d-%m-%Y', errors='coerce')
```

```python
df.head()
```

#### Convert to datetime

```python
df["TransactionDate"] = pd.to_datetime(df["TransactionDate"], errors='coerce')
```

#### Extracted the Month and the Year

```python
df['Year'] = df["TransactionDate"].dt.year.astype("Int64") #.astype("Int64") converts to integer format without decimals
df['Month'] = df["TransactionDate"].dt.month.astype("Int64")
```

```python
df.head()
```

**<h3 style="color:blue;">1.4 Ensure account types and transaction categories are standardized.</h3>**

```python
df["TransactionType"] = df["TransactionType"].str.lower().str.strip()

# Replace variations
df["TransactionType"] = df["TransactionType"].replace({
    'cr': 'credit',
    'dr': 'debit',
    'deb': 'debit'
})
```

```python
df["AccountType"] = df["AccountType"].str.lower().str.strip()

df["AccountType"] = df["AccountType"].replace({
    'sav': 'savings',
    'curr': 'current',
    'cc': 'credit'
})
```
