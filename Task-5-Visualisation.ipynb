```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_jp_morgan.csv")
```

**<h3 style="color:blue;"> 5.1 Transaction Type Distribution. </h>**

```python
transaction_counts = df["TransactionType"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(
    transaction_counts.index,
    transaction_counts.values,
    color=["#4E79A7", "#F28E2B", "#59A14F", "#E15759"]
)

plt.title("Distribution of Transaction Types", fontsize=14, fontweight="bold")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

**<h3 style="color:blue;"> 5.2 Monthly Transaction Trend. </h>**

```python
monthly_amount = (
    df.groupby("Month")["TransactionAmount"]
      .sum()
)

plt.figure(figsize=(12,5))

plt.plot(
    monthly_amount.index.astype(str),
    monthly_amount.values,
    marker="o",
    linewidth=3,
    color="#2E8B57"
)

plt.title("Monthly Transaction Amount Trend", fontsize=14, fontweight="bold")
plt.xlabel("Month")
plt.ylabel("Total Transaction Amount")

plt.xticks(rotation=45)

plt.grid(alpha=0.3)

plt.show()
```

**<h3 style="color:blue;">5.3 Average Balance by Account Type. </h>**

```python
balance = (
    df.groupby("AccountType")["AccountBalance"]
      .mean()
)

plt.figure(figsize=(7,5))

plt.bar(
    balance.index,
    balance.values,
    color="#8E44AD"
)

plt.title("Average Balance by Account Type", fontsize=14, fontweight="bold")
plt.xlabel("Account Type")
plt.ylabel("Average Balance")

plt.show()
```

**<h3 style="color:blue;"> 5.4 Top 10 Products by Transaction Volume.</h>**

```python
top_products = (
    df.groupby("Product")["TransactionAmount"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))

plt.barh(
    top_products.index,
    top_products.values,
    color="#20B2AA"
)

plt.title("Top 10 Products by Transaction Volume", fontsize=14, fontweight="bold")
plt.xlabel("Transaction Amount")
plt.gca().invert_yaxis()

plt.show()
```

**<h3 style="color:blue;"> 5.5 Regional Transaction Analysis.</h>**

```python
region = (
    df.groupby("Region")["TransactionAmount"]
      .sum()
      .sort_values()
)

plt.figure(figsize=(9,6))

plt.barh(
    region.index,
    region.values,
    color="#FF8C00"
)

plt.title("Transaction Amount by Region", fontsize=14, fontweight="bold")
plt.xlabel("Transaction Amount")

plt.show()
```

**<h3 style="color:blue;"> 5.6 Credit Rating Distribution.</h>**

```python
credit = df["CreditRating"].value_counts()

plt.figure(figsize=(8,5))

plt.pie(
    credit.values,
    labels=credit.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Credit Rating Distribution")

plt.show()
```

**<h3 style="color:blue;"> 5.7 Risk Score Distribution.</h>**

```python
plt.figure(figsize=(8,5))

plt.hist(
    df["RiskScore"],
    bins=20,
    color="#4682B4",
    edgecolor="black"
)

plt.title("Distribution of Risk Scores", fontsize=14, fontweight="bold")
plt.xlabel("Risk Score")
plt.ylabel("Frequency")

plt.show()
```

**<h3 style="color:blue;"> 5.8 Top 10 Managers by Transaction Amount. </h>**

```python
manager = (
    df.groupby("Manager")["TransactionAmount"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,5))

plt.bar(
    manager.index,
    manager.values,
    color="#CD5C5C"
)

plt.xticks(rotation=45)

plt.title("Top 10 Managers by Transaction Amount", fontsize=14, fontweight="bold")

plt.show()
```

**<h3 style="color:blue;"> 5.9 Transaction Volume by Firm. </h>**

```python
firm = (
    df.groupby("Firm")["TransactionAmount"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

plt.bar(
    firm.index,
    firm.values,
    color="#556B2F"
)

plt.xticks(rotation=45)

plt.title("Transaction Volume by Firm", fontsize=14, fontweight="bold")

plt.show()
```

**<h3 style="color:blue;"> 5.10 Top 10 Customers by Transaction Amount.</h>**

```python
top_customer = (
    df.groupby("CustomerID")["TransactionAmount"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,5))

plt.bar(
    top_customer.index.astype(str),
    top_customer.values,
    color="#D2691E"
)

plt.title("Top 10 Customers by Transaction Amount", fontsize=14, fontweight="bold")
plt.xlabel("Customer ID")
plt.ylabel("Transaction Amount")

plt.xticks(rotation=45)

plt.show()
```
